```python
import torch

data_path = "../data/filtered_train_test_split/"

POSITIONS = 18

X_train_tensor, y_train_tensor, X_test_tensor, y_test_tensor = {}, {}, {}, {}


for i in range(POSITIONS):
    X_train_tensor[i] = torch.load(data_path + f"X_train_{i}.pt")
    y_train_tensor[i] = torch.load(data_path + f"y_train_{i}.pt")
    X_test_tensor[i] = torch.load(data_path + f"X_test_{i}.pt")
    y_test_tensor[i] = torch.load(data_path + f"y_test_{i}.pt")

print(X_train_tensor[0][:5])
print(y_train_tensor[0][:5])
print(X_test_tensor[0][:5])
print(y_test_tensor[0][:5])
```

```python
from torch.utils.data import DataLoader, TensorDataset

batch_size = 512
num_workers = 16
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

inputs, train_dataset, test_dataset, train_loader, test_loader = {}, {}, {}, {}, {}

for i in range(POSITIONS):
    inputs[i] = X_train_tensor[i].shape[1]

    train_dataset[i] = TensorDataset(X_train_tensor[i], y_train_tensor[i])
    test_dataset[i] = TensorDataset(X_test_tensor[i], y_test_tensor[i])

    train_loader[i] = DataLoader(
        dataset=train_dataset[i],
        batch_size=batch_size,
        shuffle=True,
        num_workers=num_workers,
        pin_memory=True,
    )
    test_loader[i] = DataLoader(
        dataset=test_dataset[i],
        batch_size=batch_size,
        shuffle=False,
        num_workers=num_workers,
        pin_memory=True,
    )

print(train_loader[0].dataset.tensors)
print(test_loader[0].dataset.tensors)
```

```python
import os

from mlp import CsiNeuralNet

save_path = "mlp/"
num_epochs = 1000

for i in range(POSITIONS - 1, -1, -1):
    save_file = f"{save_path}filtered_model_{i}.pth"
    state_file = save_file if os.path.exists(save_file) else None
    net = CsiNeuralNet(inputs[i], device, state_file=state_file, output_dim=2)

    print(f"Training model for position {i}")
    result_model = net.train_model(
        train_loader[i], test_loader[i], save_file, num_epochs
    )
    print(f"Model for position {i} finished training")
```
